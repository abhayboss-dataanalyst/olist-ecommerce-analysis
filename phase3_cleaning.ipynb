```python
orders_df = pd.read_sql("SELECT * FROM orders", conn)
orders_df.head()
```

```python
mismatched = orders_df[(orders_df['order_status'] == 'delivered') & (orders_df['order_delivered_customer_date'].isnull())]
mismatched
```

```python
reverse_mismatch = orders_df[(orders_df['order_status'] != 'delivered') & (orders_df['order_delivered_customer_date'].notnull())]
reverse_mismatch
```

```python
# Fix the 8 rows: marked "delivered" but missing a date -> we can't invent a date,
# so we leave the date as-is but flag these for the write-up (can't safely relabel without a date)

# Fix the 6 rows: has a delivery date but status isn't "delivered" -> the date is more
# reliable evidence than the status label, so we correct the status
orders_df.loc[
    (orders_df['order_status'] != 'delivered') & (orders_df['order_delivered_customer_date'].notnull()),
    'order_status'
] = 'delivered'

# Confirm the fix worked
orders_df['order_status'].value_counts()
```

```python
orders_df['delivery_delay_days'] = (orders_df['order_delivered_customer_date'] - orders_df['order_estimated_delivery_date']).dt.days
orders_df[['order_id', 'order_delivered_customer_date', 'order_estimated_delivery_date', 'delivery_delay_days']].head(10)
```

```python
# How many days between purchase and actual delivery (total fulfillment time)
orders_df['fulfillment_days'] = (orders_df['order_delivered_customer_date'] - orders_df['order_purchase_timestamp']).dt.days

# A simple yes/no flag: was this order late?
orders_df['is_late'] = orders_df['delivery_delay_days'] > 0

orders_df[['order_id', 'fulfillment_days', 'delivery_delay_days', 'is_late']].head(10)
```

```python
order_items_df = pd.read_sql("SELECT * FROM order_items", conn)

# Combine multiple items into one row per order: total price + total freight
order_totals = order_items_df.groupby('order_id').agg(
    total_price=('price', 'sum'),
    total_freight=('freight_value', 'sum'),
    item_count=('order_item_id', 'count')
).reset_index()

order_totals.head(10)
```

```python
orders_full = orders_df.merge(order_totals, on='order_id', how='left')
orders_full.head(10)
```

```python
orders_full.isnull().sum()
```

```python
orders_full['is_late'] = (orders_full['delivery_delay_days'] > 0).astype('boolean')
orders_full.loc[orders_full['delivery_delay_days'].isnull(), 'is_late'] = pd.NA
orders_full['is_late'].value_counts(dropna=False)
```

```python
no_items = orders_full[orders_full['total_price'].isnull()]
no_items['order_status'].value_counts()
```

```python
orders_full.to_csv('orders_cleaned.csv', index=False)
```

## Data Quality Notes — Phase 3

**Status/date inconsistencies (14 orders, 0.014% of data):**
- 8 orders marked `delivered` with no `order_delivered_customer_date` recorded — left as-is, documented as a gap since no reliable date could be assigned.
- 6 orders had a delivery date recorded but were *not* marked `delivered` (mostly mislabeled as `canceled`) — corrected the status to `delivered`, since the date is stronger evidence than the label.

**Missing delivery dates (2,965 orders, ~3%):**
- These are orders that were canceled, marked unavailable, or otherwise never completed. Missing dates here are expected and meaningful, not an error — they were deliberately left as null rather than filled in.

**Orders with no associated items (775 orders):**
- 767 are `unavailable` or `canceled` orders, which makes sense — no product line was ever finalized.
- 6 are unexplained anomalies (`created`, `invoiced`, `shipped` status with zero items) — flagged but not resolved, since the count is too small to affect the overall analysis.

**Features engineered:**
- `delivery_delay_days` — actual delivery date minus estimated delivery date (negative = early, positive = late).
- `fulfillment_days` — total days from purchase to delivery.
- `is_late` — nullable boolean flag; `True`/`False` where a delivery date exists, `<NA>` where it doesn't (avoids the pandas bug where `NaN > 0` silently evaluates to `False`).
- `total_price`, `total_freight`, `item_count` — aggregated from `order_items` per order.

**Key stat:** Of orders with a known outcome, ~6.8% were delivered later than their estimated date.
